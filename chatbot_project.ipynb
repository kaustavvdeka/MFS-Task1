### Cosine Similarity Intent Detection
We test how the NLP model performs on novel phrasings that do not explicitly match keywords in the patterns.

```python
test_inputs = [
    'hi there how are you',
    'what is the scholarship amount',  # 'scholarship' is not in rule patterns
    'I need information about accommodation',  # 'accommodation' is hostel synonym
    'what jobs will I get after graduating',  # 'jobs' and 'graduating' are placement synonyms
    'I want to study computer science'  # 'computer science' is course synonym
]

print("=== NLP INTENT DETECTION TEST ===")
for inp in test_inputs:
    tag, response, score = nlp_chatbot.predict_intent_cosine(inp)
    print(f"Input: '{inp}'")
    print(f"Detected Intent: {tag} (Cosine Similarity: {score:.3f})")
    print(f"Response: {response}")
    print("-" * 60)
```

## Comparative Discussion: Rule-Based vs. NLP Chatbot

### Rule-Based Approach
- **How it works**: Compares terms literally (overlap matching) or performs search using regular expressions.
- **Strengths**: Simple to write, requires zero training data, deterministic, and highly predictable.
- **Limitations**: Breaks easily if the user phrases their query slightly differently, uses synonyms, or types a query that doesn't contain standard keywords.

### NLP-Enhanced Cosine Similarity Approach
- **How it works**: Uses Advanced preprocessing (lemmatization and stop-word filtering) to reduce noise. It maps words to TF-IDF vectors and computes the cosine of the angle between input vectors and database patterns.
- **Strengths**: Understands semantic context, generalizes to novel phrases, is robust to synonyms, and can fall back gracefully to rule-based logic when confidence is low.
- **Limitations**: Requires corpus preprocessing, relies on TF-IDF vocabulary, and might occasionally misclassify highly out-of-distribution queries without a threshold check.

### Conclusion
The **NLP-enhanced model performs significantly better** due to its ability to handle synonyms and context. By pairing it with a rule-based fallback mechanism, we create a robust, error-free chatbot engine.
